#Hill Climbing in Python — Vijaynagar → CDGI

##It demonstrates how Hill Climbing selects the neighboring location that appears closest to the goal (CDGI).

#Idea
##- Start: Vijaynagar
##- Goal: CDGI
##- h(n) = estimated distance from current location to CDGI
##- At each step, choose the neighbor with the lowest heuristic value.
##- Stop when CDGI is reached or no better neighbor exists.

```python
# Hill Climbing Search
# Example: Vijaynagar -> CDGI

# Graph: location -> neighboring locations
graph = {
    "Vijaynagar": ["Bengali Square", "MR-10"],
    "Bengali Square": ["Vijaynagar", "Rau"],
    "MR-10": ["Vijaynagar", "Bypass"],
    "Rau": ["Bengali Square", "CDGI"],
    "Bypass": ["MR-10", "CDGI"],
    "CDGI": ["Rau", "Bypass"]
}

# Heuristic values:
# Estimated distance from each location to CDGI
h = {
    "Vijaynagar": 12,
    "Bengali Square": 8,
    "MR-10": 10,
    "Rau": 4,
    "Bypass": 6,
    "CDGI": 0
}


def hill_climbing(start, goal):
    current = start
    path = [current]

    while current != goal:

        # Get neighboring locations
        neighbors = graph[current]

        # Select the neighbor with the smallest heuristic value
        best_neighbor = min(neighbors, key=lambda x: h[x])

        print(f"\nCurrent: {current}")
        print("Neighbors:")

        for neighbor in neighbors:
            print(f"  {neighbor} -> h = {h[neighbor]}")

        print(f"Selected: {best_neighbor}")

        # Hill climbing condition:
        # Move only if the neighbor is better
        if h[best_neighbor] >= h[current]:
            print("\nNo better neighboring state found.")
            print("Hill Climbing is stuck!")
            return path

        current = best_neighbor
        path.append(current)

    return path


# Run the algorithm
start = "Vijaynagar"
goal = "CDGI"

path = hill_climbing(start, goal)

print("\n----------------------------")
print("Final Path:")
print(" -> ".join(path))
print("----------------------------")
```

#Best-First Search

```python
# Graph: location -> connected locations
graph = {
    "Vijaynagar": ["Bengali Square", "MR-10"],
    "Bengali Square": ["Vijaynagar", "Rau"],
    "MR-10": ["Vijaynagar", "Bypass"],
    "Rau": ["Bengali Square", "CDGI"],
    "Bypass": ["MR-10", "CDGI"],
    "CDGI": ["Rau", "Bypass"]
}

# Heuristic values:
# Estimated distance from each location to CDGI
h = {
    "Vijaynagar": 12,
    "Bengali Square": 8,
    "MR-10": 10,
    "Rau": 4,
    "Bypass": 6,
    "CDGI": 0
}


def best_first_search(start, goal):

    # Frontier contains nodes waiting to be explored
    frontier = [start]

    # Keep track of visited locations
    visited = set()

    # Store the path
    parent = {start: None}

    while frontier:

        # Select node having the smallest heuristic value
        current = min(frontier, key=lambda x: h[x])

        # Remove current node from frontier
        frontier.remove(current)

        print(f"\nSelected: {current}")
        print(f"Heuristic h({current}) = {h[current]}")

        # Goal test
        if current == goal:
            print("\nGoal reached!")
            break

        # Mark current node as visited
        visited.add(current)

        # Explore neighboring locations
        for neighbor in graph[current]:

            if neighbor not in visited and neighbor not in frontier:

                frontier.append(neighbor)
                parent[neighbor] = current

        print("Frontier:", frontier)

    # If goal was not found
    if goal not in parent:
        return None

    # Construct path
    path = []
    current = goal

    while current is not None:
        path.append(current)
        current = parent[current]

    path.reverse()

    return path


# Start and Goal
start = "Vijaynagar"
goal = "CDGI"

# Run Best-First Search
path = best_first_search(start, goal)

# Display final path
print("\n----------------------------")
print("Final Path:")
print(" -> ".join(path))
print("----------------------------")
```
